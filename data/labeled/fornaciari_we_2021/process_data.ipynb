```python
import pandas as pd
```

```python
df = pd.read_excel("fornaciari_we_2021-pledge_annotations.xlsx")
```

```python
df.country.value_counts()
```

```python
# subset to English-language texts
df = df[df.country=='India']
```

```python
# discard the shortest and longest texts, which are outliers
df['len_quant'] = pd.qcut(df.len, q=8, labels=False)
df = df[df.len_quant!=df.len_quant.min()]
df = df[df.len_quant!=df.len_quant.max()]
```

```python
# inspect the data split distribution
df['set'].value_counts()
```

```python
cols_map = {
    'manifesto_id': 'text_id',
    'clean_text': 'text',
    'binary_pledge': 'label',
    'party': 'metadata__party',
    'year': 'metadata__year',
    'set': 'metadata__split'
}
out = df[list(cols_map.keys())].rename(columns=cols_map)
```
