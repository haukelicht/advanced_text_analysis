### Setup

```python
import os
import openai
from dotenv import load_dotenv
print("openai version:", openai.__version__) # should be ~3.8.0
from openai import OpenAI

SEED = 42
```

```python
# access the environment variable OPENAI_KEY and set it as the API key
load_dotenv()
client = OpenAI(
  api_key=os.environ["OPENAI_API_KEY"],  # this is also the default, it can be omitted
)
```

## Tests

### Chat completion models

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Hello, who won the Nobel Prize in Physics in 2021?"}
]
```

#### GPT-4o

all available models are listed here: https://developers.openai.com/api/docs/models/all#all

```python
try:
    # Make a request to the GPT-4o chat completions endpoint
    response = client.chat.completions.create(
        model="gpt-4o",
        messages=messages, 
        max_tokens=5,
        seed=SEED
    )
    
except Exception as e:
    print("The API key is not working")
    print(f"An error occurred: {e}")
else:
    print(f"The API key is working! response: '{response.choices[-1].message.content}'")
```
